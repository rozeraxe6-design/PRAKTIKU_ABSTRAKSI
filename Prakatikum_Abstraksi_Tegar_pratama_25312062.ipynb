````python
from abc import ABC, abstractmethod

class kendaraan(ABC):
  def __init__(self, nama):
    self.nama = nama

  def buat_info(self):
    return f"nama kendaraan {self.nama}"

  @abstractmethod
  def bergerak(self):
    pass

class mobil(kendaraan):
  def __init__(self, nama, jumlah_pintu=0):
    super().__init__(nama)
    self.jumlah_pintu = jumlah_pintu

  def bergerak(self):
    return f"{self.nama} motor bergerak dengan kecepatan tinggi"

class motor(kendaraan):
  def bergerak(self):
    return f"{self.nama} motor bergerak dengan kecepatan roda"

class sepeda(kendaraan):
  def bergerak(self):
    return f"{self.nama} sepeda bergerak dengan kecepatan rendah"

#fungsi panggilan hanya membutuhkan antasr muka yang sesuai

def jalankan_kendaraan(kendaraan):
  print(kendaraan.buat_info())
  print(kendaraan.bergerak())

mobil_baru = mobil("mobil baru", 4)
motor_baru = motor("motor baru")
sepeda_baru = sepeda("sepeda baru")

semua_kendaraan = [mobil_baru, motor_baru, sepeda_baru]
for kendaraan in semua_kendaraan:
  jalankan_kendaraan(kendaraan)

print("jumlah pintu mobiol;", mobil_baru.jumlah_pintu)
````

## 1. Apa itu abstraksi?
**Abstraksi memilih fitur penting dan menyediakan antarmuka yang sederhana**, sehingga pemakai tidak perlu bergantung pada semua rincian implementasi. Contoh: pemanggil cukup menjalankan kendaraan.bergerak(), sedangkan cara bergeraknya ditentukan masing-masing jenis kendaraan.

Class abstrak adalah salah satu cara menerapkan prinsip itu di Python. Abstraksi juga dapat dicapai melalui fungsi dan antarmuka sederhana tanpa harus selalu memakai ABC.

| Konsep | Fokus pada contoh |
|---|---|
| Enkapsulasi | Mengelompokkan keadaan dan perilaku, serta mengatur cara data diperbarui; misalnya nilai melalui setter |
| Abstraksi | Menyediakan operasi bergerak() yang perlu diketahui pemanggil, tanpa harus mengetahui isinya |
| Pewarisan | Mobil dan Motor merupakan jenis Kendaraan |
| Polimorfisme | Perintah bergerak() yang sama menghasilkan aksi sesuai objek |

Konsep tersebut saling melengkapi. Atribut __ bukan syarat abstraksi dan bukan pengamanan mutlak; penamaan internal serta validasi data perlu dirancang tersendiri.

## 2. Class abstrak, method abstrak, dan decorator
- **ABC** adalah class bantuan dari modul abc untuk mendefinisikan abstract base class.
- **@abstractmethod** menandai method yang masih menjadi kewajiban implementasi. Dengan mekanisme ABC, class yang masih memiliki method abstrak belum dapat dibuat objeknya.
- **Class konkret** sudah memenuhi seluruh kewajiban method abstrak yang diwarisi, sehingga dapat diinstansiasi.
- Class abstrak dapat memiliki __init__, atribut, dan method biasa yang isinya sudah lengkap.

**Nuansa penting:** turunan boleh belum mengimplementasikan method abstrak, tetapi turunan itu masih abstrak. Implementasi juga dapat diwarisi dari turunan perantara yang sudah konkret; tidak harus ditulis ulang di setiap class paling bawah.

# 3. Praktik bersama — Satu program kendaraan
Kendaraan menentukan antarmuka bergerak(). Mobil, Motor, dan Sepeda mengimplementasikannya. Fungsi jalankan_kendaraan() cukup memakai antarmuka tersebut tanpa percabangan berdasarkan jenis kendaraan.

**Prediksi sebelum menjalankan:** apakah nama method yang dipanggil sama? Mengapa keluarannya berbeda?

````python
from abc import ABC, abstractmethod


# CLASS ABSTRAK: menentukan antarmuka bersama
class Kendaraan(ABC):
    def __init__(self, nama):
        self.nama = nama

    def buat_info(self):
        return f"Kendaraan: {self.nama}"

    @abstractmethod
    def bergerak(self):
        pass


# CLASS KONKRET: memenuhi method abstrak
class Mobil(Kendaraan):
    def __init__(self, nama, jumlah_pintu):
        super().__init__(nama)
        self.jumlah_pintu = jumlah_pintu

    def bergerak(self):
        return f"{self.nama} melaju di jalan dengan empat roda."


class Motor(Kendaraan):
    def bergerak(self):
        return f"{self.nama} melaju di jalan dengan dua roda."


class Sepeda(Kendaraan):
    def bergerak(self):
        return f"{self.nama} bergerak ketika pedal dikayuh."


# FUNGSI PEMANGGIL: hanya membutuhkan antarmuka yang sesuai

def jalankan_kendaraan(kendaraan):
    print(kendaraan.buat_info())
    print(kendaraan.bergerak())
    print()


# MEMBUAT OBJEK DARI CLASS KONKRET
mobil1 = Mobil("Mobil Kampus", 4)
motor1 = Motor("Motor Mahasiswa")
sepeda1 = Sepeda("Sepeda Hijau")

semua_kendaraan = [mobil1, motor1, sepeda1]
for kendaraan in semua_kendaraan:
    jalankan_kendaraan(kendaraan)

print("Jumlah pintu mobil:", mobil1.jumlah_pintu)
````

## 4. Penjelasan setiap bagian kode
| Baris / bagian | Penjelasan |
|---|---|
| from abc import ABC, abstractmethod | Mengambil dua fasilitas abstraksi dari modul standar Python. |
| class Kendaraan(ABC) | Menyiapkan class yang dapat menerapkan kewajiban method abstrak. Mewarisi ABC saja tanpa method abstrak tidak otomatis melarang pembuatan objek. |
| def __init__(self, nama) | Menginisialisasi nama pada objek turunan melalui initializer induk. |
| self.nama = nama | Menyimpan nilai parameter nama sebagai atribut objek. |
| def buat_info(self) | Method biasa yang implementasinya sudah tersedia; dipakai kembali oleh semua turunan. |
| return f"Kendaraan: ..." | Menyusun teks dari atribut nama dan mengembalikannya kepada pemanggil. |
| @abstractmethod | Decorator yang menandai bergerak() sebagai method abstrak dalam mekanisme ABC. |
| def bergerak(self): pass pada induk | Menyatakan antarmuka. Pass adalah isi kosong pada contoh ini, bukan penyebab class menjadi abstrak. |
| class Mobil(Kendaraan) | Membuat turunan dari Kendaraan. |
| __init__(self, nama, jumlah_pintu) | Menerima data khusus Mobil selain nama. |
| super().__init__(nama) | Menggunakan initializer induk pada objek Mobil yang sama. |
| self.jumlah_pintu = jumlah_pintu | Menambahkan atribut khusus Mobil. |
| bergerak() pada Mobil | Mengimplementasikan kewajiban abstrak dan meng-override method induk. |
| class Motor / Sepeda dan bergerak() | Mengimplementasikan operasi sama dengan isi berbeda. |
| Motor("Motor Mahasiswa") | Memakai __init__ induk karena Motor tidak mendefinisikan initializer sendiri. Sepeda juga begitu. |
| def jalankan_kendaraan(kendaraan) | Fungsi menerima satu objek yang menyediakan method yang dipakai. |
| print(kendaraan.buat_info()) | Menampilkan informasi umum dari method biasa induk. |
| print(kendaraan.bergerak()) | Memanggil implementasi konkret sesuai objek saat program berjalan. |
| print() | Memberi pemisah tampilan. |
| Tiga pembuatan objek | Membuat instance Mobil, Motor, dan Sepeda; bukan instance langsung Kendaraan. |
| semua_kendaraan = [...] | List yang menampung objek berbeda. |
| for kendaraan in semua_kendaraan | Menjalankan fungsi yang sama untuk setiap objek. |
| print jumlah_pintu | Membuktikan objek Mobil juga mempunyai atribut khusus. |

**Letak abstraksi:** pemanggil memakai bergerak() tanpa perlu mengetahui isi setiap class. **Letak polimorfisme:** pemanggilan itu menghasilkan perilaku berbeda sesuai objek. **Letak kontrak ABC:** objek class yang masih memiliki method abstrak tidak dapat dibuat.

## 5. Apa fungsi simbol @ dan decorator?
Decorator diterapkan pada fungsi atau class yang berada di bawahnya. Ia dapat menandai atau mengubah perilaku objek yang didekorasi. Sebagian decorator membungkus fungsi dengan perilaku tambahan; @abstractmethod pada contoh ini terutama **memberi penanda abstrak** yang diperiksa oleh mekanisme ABC.

```python
@abstractmethod
def bergerak(self):
    pass
```

Tulis @abstractmethod tepat di atas method yang ditandai. __init__ dan buat_info() pada contoh tidak abstrak sehingga tidak memakai penanda itu. Decorator tidak harus selalu berkaitan dengan abstraksi; misalnya decorator lain dapat dipakai untuk pencatatan atau pengaturan akses.

ABC tidak otomatis membuktikan isi method sudah benar atau memeriksa semua kesesuaian parameter. Pengembang tetap harus menjaga antarmuka bergerak(self) dan perilakunya agar dapat dipakai pemanggil.

## 6. Mengamati class abstrak tanpa menghentikan notebook
Sel berikut memeriksa status class dengan inspect.isabstract(). KendaraanBelumLengkap belum mengimplementasikan bergerak(), sehingga tetap abstrak. Sel ini tidak mencoba membuat objek dari class yang belum lengkap.

````python
from inspect import isabstract


class KendaraanBelumLengkap(Kendaraan):
    pass


print("Kendaraan abstrak:", isabstract(Kendaraan))
print("Mobil abstrak:", isabstract(Mobil))
print("Turunan belum lengkap abstrak:", isabstract(KendaraanBelumLengkap))

# Opsional: hapus tanda # pada SATU baris untuk melihat TypeError.
# objek_abstrak = Kendaraan("Contoh")
# objek_belum_lengkap = KendaraanBelumLengkap("Contoh")
````

Hasil yang diharapkan: **True, False, True**. isabstract() hanya memeriksa status class; bukan penanganan error.

Jika mencoba membuat Kendaraan(...) atau KendaraanBelumLengkap(...), Python menampilkan **TypeError** karena kewajiban bergerak() belum dipenuhi. Jika menjalankan percobaan opsional, kembalikan tanda komentar sesudah mengamati pesannya agar Jalankan semua tetap lancar.

**Bandingkan dengan pertemuan pewarisan:** class induk biasa bisa menyediakan bergerak() umum dan dapat dibuat objeknya. Pada class Kendaraan yang memakai @abstractmethod, kita sengaja menetapkan bahwa jenis kendaraan konkret perlu menyediakan implementasi dahulu.

````python
from abc import ABC, abstractmethod
from inspect import isabstract


# CLASS ABSTRAK: menentukan antarmuka bersama
class Kendaraan(ABC):
    def __init__(self, nama):
        self.nama = nama

    def buat_info(self):
        return f"Kendaraan: {self.nama}"

    @abstractmethod
    def bergerak(self):
        pass

    @abstractmethod
    def sumber_energi(self):
        pass


# CLASS KONKRET: memenuhi seluruh method abstrak
class Mobil(Kendaraan):
    def __init__(self, nama, jumlah_pintu):
        super().__init__(nama)
        self.jumlah_pintu = jumlah_pintu

    def bergerak(self):
        return f"{self.nama} melaju di jalan dengan empat roda."

    def sumber_energi(self):
        return "Bensin"


class Motor(Kendaraan):
    def bergerak(self):
        return f"{self.nama} melaju di jalan dengan dua roda."

    def sumber_energi(self):
        return "Bensin"


class Sepeda(Kendaraan):
    def bergerak(self):
        return f"{self.nama} bergerak ketika pedal dikayuh."

    def sumber_energi(self):
        return "Tenaga manusia"


# TUGAS 1: class Bus dengan kapasitas_penumpang
class Bus(Kendaraan):
    def __init__(self, nama, kapasitas_penumpang):
        super().__init__(nama)
        self.kapasitas_penumpang = kapasitas_penumpang

    def bergerak(self):
        return (f"{self.nama} mengangkut penumpang di jalan "
                f"dengan kapasitas {self.kapasitas_penumpang} orang.")

    def sumber_energi(self):
        return "Diesel"


# FUNGSI PEMANGGIL (TUGAS 2): diubah satu kali, tanpa if/elif jenis kendaraan
def jalankan_kendaraan(kendaraan):
    print(kendaraan.buat_info())
    print(kendaraan.bergerak())
    print("Sumber energi:", kendaraan.sumber_energi())
    print()


# TUGAS 3: satu objek setiap jenis dalam satu list
mobil1 = Mobil("Mobil Kampus", 4)
motor1 = Motor("Motor Mahasiswa")
sepeda1 = Sepeda("Sepeda Hijau")
bus1 = Bus("Bus Kampus", 40)

semua_kendaraan = [mobil1, motor1, sepeda1, bus1]
for kendaraan in semua_kendaraan:
    jalankan_kendaraan(kendaraan)

print("Jumlah pintu mobil:", mobil1.jumlah_pintu)
print("Kapasitas bus:", bus1.kapasitas_penumpang)

# Pemeriksaan status abstrak
print("Kendaraan abstrak:", isabstract(Kendaraan))
print("Mobil abstrak:", isabstract(Mobil))
print("Bus abstrak:", isabstract(Bus))
````
